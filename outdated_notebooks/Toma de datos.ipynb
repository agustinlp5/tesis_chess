```python
import zstandard as zstd
import chess.pgn
import io
MAX_GAMES = 100000
```

```python
input_file = "c:/Users/agust/Desktop/facu temporal/tesis/lichess_db_standard_rated_2019-01.pgn.zst"

output_file = "C:/Users/agust/Desktop/facu temporal/tesis/generado/partidas.pgn"
```

```python
def parse_time_control(tc):
    """Devuelve el tiempo base en segundos o 0 si hay error."""
    try:
        base = int(tc.split("+")[0])
        return base
    except:
        return 0


def partida_valida(headers):
    try:
        white_elo = int(headers.get("WhiteElo", 0))
        black_elo = int(headers.get("BlackElo", 0))

        time_control = headers.get("TimeControl", "")

        if white_elo < 900 or black_elo < 900:
            return False

        base_seconds = parse_time_control(time_control)

        if base_seconds < 180:
            return False

        return True

    except:
        return False
```

```python
with open(input_file, "rb") as compressed, \
     open(output_file, "w", encoding="utf-8") as out:

    dctx = zstd.ZstdDecompressor()

    with dctx.stream_reader(compressed) as reader:

        text_stream = io.TextIOWrapper(reader, encoding="utf-8")

        count_in = 0
        count_valid = 0

        while count_valid < MAX_GAMES:

            game = chess.pgn.read_game(text_stream)

            count_in += 1

            if game is None:
                break

            if partida_valida(game.headers):

                # Guardar partida
                out.write(str(game) + "\n\n")

                count_valid += 1

            if count_in % 1000 == 0:
                print(
                    f"Procesadas {count_in} partidas, "
                    f"guardadas {count_valid} partidas válidas."
                )

        print(
            f"Procesadas {count_in} partidas. "
            f"Guardadas {count_valid} partidas válidas."
        )
```
